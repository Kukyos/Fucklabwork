### NumPy

```python
import numpy as np
np.random.seed(1021)

# hourly defect rates for a full day (24 values, 0.1 - 5.0 %)
defects = np.round(np.random.uniform(0.1, 5.0, 24), 2)
defects
```

```python
# defect rate at the 12th hour
defects[12]
```

```python
# night shift runs 20:00 to 06:00, so it wraps past midnight
night = np.concatenate((defects[20:], defects[:7]))
night
```

```python
# 3 shifts of 8 hours each
shifts = defects.reshape(3, 8)
shifts
```

```python
# hours where defects went above 3%
for hour, rate in enumerate(defects):
    if rate > 3:
        print("hour", hour, "->", rate, "%")
```

```python
# pair each hour's defects with a temperature reading
temp = np.round(np.random.uniform(18, 30, 24), 1)
combined = np.vstack((defects, temp))
combined
```

```python
s1, s2, s3 = np.split(defects, 3)
print("shift 1:", s1)
print("shift 2:", s2)
print("shift 3:", s3)
```

```python
# anything above 4% counts as critical
defects[defects > 4]
```

```python
np.sort(defects)
```

```python
# acceptable quality is between 0.5 and 1.5%
defects[(defects >= 0.5) & (defects <= 1.5)]
```
