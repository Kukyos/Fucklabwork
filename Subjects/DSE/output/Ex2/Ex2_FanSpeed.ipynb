```python
print("URK24CS1021")

df["Speed_Per_Degree"] = df["Observed Fan Speed"] / df["Temperature"]
df["Speed_Level"] = df["Observed Fan Speed"].apply(lambda x: "High" if x > 60 else "Low")
df.head()
```

```python
print("URK24CS1021")

df.sort_values("Observed Fan Speed", ascending=False).head()
```

```python
print("URK24CS1021")

df.groupby("Speed_Level")["Observed Fan Speed"].mean()
```

```python
print("URK24CS1021")

df["Rank_dense"] = df["Observed Fan Speed"].rank(method="dense")
df.sort_values("Rank_dense").head()
```
